```python
import numpy as np
import plotly.express as px

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

fig = px.line(x=x, 
              y=y, 
              title="It Works",
              labels={
                  "x" : "Input",
                  "y" : "sin(x)"
              })

fig.show()
```
